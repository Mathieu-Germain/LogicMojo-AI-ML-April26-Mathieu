**Copying**

Understanding how to duplicate data without affecting the original.

copy(): Create original_list = [1, 2, 3]. Create a new_list using the copy method. Append 4 to the new_list. Check both lists—did the original_list change?

```python
original_list = [1, 2, 3]

print(original_list)

new_list = original_list.copy()

print(new_list)

new_list.append(4)

print(new_list)
print(original_list)

print()

#copy using : (slice) operator

new_list2 = original_list[:]
print(new_list2)

print(original_list[:])

new_list2.append(5)
print(new_list2)

print(original_list)
```

***Pro-Tip: The "None" Trap***

A common mistake when starting out is writing code like this:
my_list = [3, 1, 2].sort()

If you print my_list, it will be None! This is because methods like .sort(), .append(), and .reverse() modify the list in-place and return nothing. Always call the method on one line, and use the list on the next.

**1. The Membership Cleanse**

Concept: in operator, loops, and list creation.


- Challenge: You are given a list of raw transaction values: transactions = [10.50, "ERROR", 25.00, "ERROR", 4.75, "PENDING", 100.00]. Create a new list called cleaned_transactions that only contains the valid float numbers.


- Hint: Use the type() function or check if the element is an instance of float.

```python
transactions = [10.50, "ERROR", 25.00, "ERROR", 4.75, "PENDING", 100.00]
# Initialize an empty list for the new transactions
new_transactions = []

#Loo through each transaction
for item in transactions:

  # check if the item is specifically a float
  if type(item) is float:
    new_transactions.append(item)

# print the result
print(new_transactions)

```

```python
# Alternative Pythonic Way: List Comprehension

transactions = [10.50, "ERROR", 25.00, "ERROR", 4.75, "PENDING", 100.00]

new_transactions = [item for item in transactions if type(item) is float]

cleaned_transactions = [item for item in transactions if isinstance(item, float)]
print(new_transactions)

```

**2. The Alternating Merger**

Concept: Loops and indexing.

- Challenge: You have two lists of equal length: letters = ['a', 'b', 'c'] and numbers = [1, 2, 3]. Merge them into a single list called combined where the elements alternate: ['a', 1, 'b', 2, 'c', 3].

- Bonus: Try using Python's built-in zip() function alongside a loop to make this even cleaner.
- Since both lists are of equal length, you can use a single for loop with an index (i) to grab the element at that position from both lists consecutively

```python
# Method 1: Using range() and Indexing
letters = ['a', 'b', 'c']
numbers = [1, 2, 3]

# initialize the empty list
combined = []

#loop throught the indices of the lists
for i in range(len(letters)):
  combined.append(letters[i])
  combined.append(numbers[i])

print(combined)
```

```python
# Method 2: The Cleaner Way (Using zip())
letters = ['a', 'b', 'c']
numbers = [1, 2, 3]

combined = []

# zip() pairs them up, and we unpack them into 'letter' and 'number'
for letter, number in zip(letters, numbers):
    combined.append(letter)
    combined.append(number)

print(combined)

```

```python
# Method 3: The Cleaner Way (Using zip())
letters = ['a', 'b', 'c']
numbers = [1, 2, 3]

# Pro-Tip (The Ultra-Concise Way):
## If you ever want to do this in a single line without an explicit loop,
## you can combine zip() with a list comprehension, like this:

combined = [item for pair in zip(letters, numbers) for item in pair]

print(f'the combined is: {combined}')

```

3.** The Target Sublist Locator**

-Concept: Lookahead indexing with index().

- Challenge: Given the list items = ["apple", "banana", "orange", "banana", "grape"], find the index of the second occurrence of "banana".

- Hint: Look closely at the optional documentation for the .index() method. It allows you to specify a start position for your search.

```python
items = ["apple", "banana", "orange", "banana", "grape"]

first_banana = items.index('banana')

second_banana = items.index('banana', first_banana + 1)


print(f'The first banana is at index {first_banana}.')
print(f'The second banana is at index {second_banana}.')
```

**The Secret to .index()**

- By default, items.index("banana") scans the list from the very beginning (index 0) and stops the moment it finds the first match.

- However, .index() accepts optional start and end arguments:
list.index(element, start, end)

- To find the second occurrence, you just need to find the first one, add 1 to its position, and tell Python to start searching from there.

**4. Chunking a List**
Concept: Advanced slicing.

- Challenge: Take a flat list of numbers, data = [1, 2, 3, 4, 5, 6, 7, 8, 9], and split it into smaller sublists of size 3. Your final output should look like a nested list: [[1, 2, 3], [4, 5, 6], [7, 8, 9]].

- Hint: Use a while loop or a range() loop with a step size of 3, slicing the list along the way.

```python
# Method 1: Using range() with a Step Size

data = [1, 2, 3, 4, 5, 6, 7, 8, 9]
chunk_size = 3

# Initialize an empty list to store the chunks
chunked_list = []

# Iterate through the list with a step size of chunk_size
for i in range(0, len(data), chunk_size):
  # Slice the list to get a chunk and append it to chunked_list
  chunked_list.append(data[i:i + chunk_size])

print(chunked_list)

```

```python
# Method 2: The Pythonic One-Liner (List Comprehension)

data = [1, 2, 3, 4, 5, 6, 7, 8, 9]
chunk_size = 3

# Combine the loop and slicing into a single line
chunked_data = [data[i : i + chunk_size] for i in range(0, len(data), chunk_size)]

print(chunked_data)
```

**The Palindrome Checker**

Concept: Reversing and slicing syntax.

- Challenge: Write a small script that takes a list of elements—for example, chars = ['r', 'a', 'd', 'a', 'r']—and checks if the list reads the same forward and backward. Print True if it is a palindrome, and False otherwise.

- Constraint: Try to do the actual check in a single line of code using a slice trick ([::-1]).

```python
# chars = ['r', 'a', 'd', 'a', 'r']

chars = input(f'Please enter you word to check: ')

# The single-line palindrome check using the slice trick
is_palindrome = chars == chars[::-1]

print(is_palindrome)
```

```python
chars = input(f'Please enter you word to check: ')

if chars == chars[::-1]:
  print(f'Congrats {chars} is a palindrome')
else:
    print(f'Congrats {chars} is not palindrome')
```

**6. Swapping Extreme Values**

Concept: min(), max(), and element swapping.

- Challenge: Given the list scores = [45, 82, 99, 12, 67], write code that finds the maximum value and the minimum value, and swaps their positions in the list. The final list should look like [45, 82, 12, 99, 67].

- Hint: Use .index() to find out where the min and max values live before you try to swap them.


```python
scores = [45, 82, 99, 12, 67]

max_score = max(scores)
min_score = min(scores)

print(f'The maximum value is {max_score}.')
print(f'The minimum value is {min_score}.')

max_index = scores.index(max_score)
min_index = scores.index(min_score)

print(f'The maximum value index is {max_index}.')
print(f'The minimum value index is {min_index}.')

print()
scores[max_index], scores[min_index] = scores[min_index], scores[max_index]

print(scores)

#not advisibe it is a static solution
scores[2:4] = [scores[2], scores[3]]
print(scores)
```
