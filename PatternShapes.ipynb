Creating pattern in Python: Increasing Triangle

In Python, end= is a special keyword used inside the print() function to control what happens at the end of the line.
Here is the simplest explanation:

**Default Behavior**: Normally, print() ends with a hidden "new line" character (\n), which moves the cursor to the next line automatically.

**What end= does:** It allows you to replace that automatic new line with something else (a space, a comma, or nothing at all).

**Note: in VS code the shortcut to copy down is Shft + Alt + Down Arrow **





```python
n = int(input("Please enter the number of rows to be displayed "))
for i in range(n):
  for j in range(i+1):
    print("* ", end="")
  print("")
```

Creating pattern in Python: decreaasing Triangle Pattern

```python
n = int(input("Please enter the number of rows to be displayed "))
for i in range(n):
  for j in range(n -i):
    print("* ", end= "")
  print()
```

```python
n = int(input("Please enter the number of rows to be displayed "))
for i in range(n,0,-1):
  for j in range(i,0,-1):
    print("*", end =" ")
  print()
```

PRINTING THE HOLLOW O SHAPE;

```python
rows = int(input("Please enter the number of rows to be displayed "))
cols = int(input("Please enter the number of columns to be displayed "))
for row in range(rows):
  for col in range(cols):
    if ((col == 0 or col == 4) and (row != 0 and row!=6)) or  ((row == 0 or row == 6) and (col > 0 and col < 4)):
      print("*", end="")
    else:
      print(end=" ")
  print()
```
