Given a list of numbers, find the second largest number in the list.

Note: There might be repeated numbers in the list.
 If there is only one number present in the list, return 'not present'.

Examples:
Input 1:
[7, 2, 0, 9, -1, 8]
Output 1:

```python
def get_second_largest(nums):

    unique_nums = list(set(nums))


    if len(unique_nums) < 2:
        return 'not present'


    unique_nums.sort(reverse=True)


    return unique_nums[1]

print(f"The second largest number from the List [10, 20, 4, 45, 99, 99] is : {get_second_largest([10, 20, 4, 45, 99, 99])}")
print(f"List [10, 10, 10]: {get_second_largest([10, 10, 10])}")
print(f"List [5]: {get_second_largest([5])}")
```

```python
def get_second_largest(nums):
    # 1. Remove duplicates by converting the list to a set
    unique_nums = set(nums)

    # 2. Check if we have at least two unique numbers
    if len(unique_nums) < 2:
        return 'not present'

    # 3. Sort the unique numbers into a list
    sorted_nums = sorted(list(unique_nums))

    # 4. Return the second to last element
    return sorted_nums[-2]

# Test Input
input_list = [7, 2, 0, 9, -1, 8]
print(get_second_largest(input_list))
```

**Get a set of numbers from the user and convert it into a list**

To get a set of numbers from a user and convert it into a list:
You need to handle three steps:

1)- taking the raw string input,

2)- cleaning/splitting that string into individual numbers, and

3)- then performing the type conversion.



```python
# 1. Get input from the user (e.g., "10 20 30 40")
user_input = input("Enter a set of numbers separated by spaces: ")

# 2. Split the string into a list of strings and convert each to an integer
# Using a set first removes any duplicates provided by the user
number_set = {int(x) for x in user_input.split()}

# 3. Convert the set into a list
number_list = list(number_set)

print(f"The resulting list is: {number_list}")
print(f"The data type is now: {type(number_list)}")
```
