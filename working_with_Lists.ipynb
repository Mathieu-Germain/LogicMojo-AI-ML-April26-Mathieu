LIST: INDEXING AND SLICING - ACCESS & READ

```python
#INDEXING:
lst = ['a', 'b', 'c', 'd']
print(lst)

print()

#Accession list using index

print(lst[0]) #will print the fist item

print()

print(lst[-1]) # will print the last item
print()


matrix = [
    ['a', 'b', 'c'],
    ['d', 'e', 'f'],
    ['g', 'h', 'i']
]

print(matrix)

print(matrix[0]) #print the first row

print()

print(matrix[-1]) #print the last row

print()

print(matrix[-1][-1]) #print the last row the last item

print()

print(matrix[0][0]) #print the item of lthe first row


print()

print(matrix[1][1]) #print the second item of the second row == 'e

print()
print(matrix[2][:2]) #print the first 2 items from row 3

print()
print(matrix[1][-1]) #print the last item of the second row == 'e

print()

print(matrix[:2]) #print the first two rows ==

print()

```

```python
#SLICING:
## Syntax [start: end] - start: included & end: excluded
lst = ['a', 'b', 'c', 'd']

print(lst[:]) # print all

print()


print(lst[0:]) # print print from first index to the last

print()


print(lst[2:]) # print from the 3rd item to the last

print()


print(lst[:-1]) # print from the last item to 1st

print()


print(lst[:-2]) # print all

print()


print(lst[:-3]) # print all

print()


```
