========== LOC & ILOC ===================

The easiest way to remember the difference is by looking at the letters in their names:

- loc is for label-based indexing (using the names of rows or columns).

- iloc is for integer-based indexing (using the 0-indexed position, just like a standard Python list).

To use the 'Roll No' column as the index, you should use df.set_index('Roll No')

```python
# LOC ## Let change our index to Rll No. Notice that it start a 1 not 0
df = df.set_index('Roll No')
df.head()
```

```python
df.loc[1] # Our index row start at 1 not 0
```

```python
df.loc[5]
```

```python
# We can provide array of index:
df.loc[[3,5,2]]
```

```python
# to see the physics mark for index 5
df.loc[5, 'Physics']
```

**The Big Gotcha: Slicing Behavior ⚠️**

The most common mistake people make comes when slicing multiple rows (using the : operator). They handle the stop value differently:

- loc includes the last item. (It's inclusive).

- iloc excludes the last item. (It behaves exactly like standard Python lists).

```python
# Slicing ==> 1:5. To see the Physics mark from 1 to 5.  includ 5 with iloc it will exclud 5

df.loc[1:5, 'Physics']
```

```python
# To see the Physics marks <50
df_50 = df.loc[df['Physics']< 50]
df_50
```

```python
# show only the maths result for student with > 80 in Physics
df.loc[df['Physics']>80, ['Maths']]
```

```python
# With iloc to see the first index [0]

df.iloc[0]

```

```python
df.iloc[1:5, 0] # 0 here is teh index position of Physics
```
