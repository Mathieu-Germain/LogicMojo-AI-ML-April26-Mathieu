CREATING DICTIONARY - Methods:

```python
user = {"id" :1, "age": 30, "city": "Yaounde"}
#Access using a key thatis not available this give an error. we can access using the get():
#print(user["name"]) # ==> KeyError

print(user.get("age")) # this return the value 30 if the value is absent it's return None

print(user.get("name"))
# return a default value when not available:
print(user.get("name", "not Available"))

```

```python
# to check is the key exist or is part of th dictionary:
print("age" in user) # this returns True.
print("name" in user) # this returns False.
print("name" not in user) # this returns True.

#View Objects:
## Keys
print(user.keys()) #==> Print Keys
print(user.values()) #==> Print values
print()
print(user.items()) #==> Print items() best used when need for looping, transforming data, buildimg new dicts, comparing and ...

print()

for u in user:
  print(u, user[u]) #==> return value but way is as follow:
print()
print()
for key, value in user.items():
  print(key, value) #==> return the samething as above.
```

ADD, UPDATE, UPDATE, DELETE VALUE IN A DICTIONARY

```python
user = {"id" :1, "age": 30, "city": "Yaounde"}

#TO ADD:
user["name"] = "Ted"
print(user)
print()

#TO UPDATE A VALUE: AGE TO 50
user["age"]=50 # this update the age to 50
print(user)

print()

#to update multiple values:
user.update({"age": 40, "city": "douala"})
print(user)

print()

#TO REMOVE USE THE POP():
user.pop("age") # removes age user pair
print(user)

print()

city = user.pop("city")
print(user)
print("Removed Item is: ", city)

city = user.pop("city", "Not Found" ) # assigning a default value when no found
print(user)
print("Removed Item is: ", city)


user.popitem() # remove the last item
print(user) # popitem() remove the last item pairs of the dictionary
```

CHALLENGE: user - {"id": 1, "name": John, "age": 49, "city": "yaounde"}

1)- create a new dictionary

2)- Keep ONLY PAIRS with string values

3)- Convert Values to Uppercase

Here we need to do 2 things:
* filterering because we want to keep only string valueto
* Transformation because we are transforming into uppercase

 to do that we use Dict Comprehension:
 - Expression
 - Loop
 - Filter




```python
#SOLUTION:

user = {"id": 1, "name": "John", "age": 49, "city": "yaounde"}
user_str ={
  k: v.upper() #Expression

  for k, v in user.items()  #Loop

  if isinstance(v, str) #Filter
}
print(user_str)
```
