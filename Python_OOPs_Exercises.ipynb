Think of it like the command "Speak":

A Dog speaks by barking.
A Car "speaks" by honking.
The command is the same, but the action depends on who is receiving the comman

🔄 The Magic of the Loop
The coolest part of polymorphism is that you can put different objects into a list and treat them all the same. You don't need to know if it's a truck or a bike; you just tell it to start_engine().


```python
class Vehicle:
    def __init__(self, make, year):
        self.make = make
        self.year = year

    def display_info(self):
        print(f"Make: {self.make}, Year: {self.year}")

    def start_engine(self):
        print("Engine starting...")

class ElectricCar(Vehicle):
    def __init__(self, make, year, battery_size):
        super().__init__(make, year)
        self.battery_size = battery_size

    def display_info(self):
        super().display_info()
        print(f"Battery Size: {self.battery_size} kWh")

    def start_engine(self):
        print("... (Silence) ... The Electric Car is ready to silent-drive.")

class GasTruck(Vehicle):
    def __init__(self, make, year, towing_capacity):
        super().__init__(make, year)
        self.towing_capacity = towing_capacity

    def display_info(self):
        super().display_info()
        print(f"Towing Capacity: {self.towing_capacity} lbs")

    def start_engine(self):
        print("VROOM!! The heavy truck engine roars to life.")

class Motorcycle(Vehicle):
    def __init__(self, make, year):
        super().__init__(make, year)

    def start_engine(self):
        print("Pop-pop-pop! The bike engine is idling.")

# Create a list (a "fleet") of different objects
my_fleet = [
    ElectricCar("Tesla", 2023, 75),
    GasTruck("Ford", 2022, 5000),
    Motorcycle("Harley", 2020)
]

# Polymorphism in action:
for vehicle in my_fleet:
    vehicle.display_info() # Shared from Parent
    vehicle.start_engine() # Unique to each Child
    print("-" * 10)
```

```python
Encapsulation is about keeping data safe by "locking" it inside the class. It prevents people from accidentally changing important values from the outside.
Think of it like a Bank Account:

- You can't just reach into the vault and change your balance.
- You have to use a specific method (like deposit or withdraw) that follows the bank's rules.


🔒 How to Lock Data in Python:
In Python, we use two underscores __ before an attribute name to make it "Private."

- self.brand → Public (Anyone can see or change it).

- self.__mileage → Private (Only the class itself can see or change it).

📝 Example: The Secure Odometer
Let's protect a car's mileage so people can't just set it back to zero to scam buyers!
```

```python
class Car:
    def __init__(self, brand):
        self.brand = brand
        self.__mileage = 0  # Private attribute

    def drive(self, miles):
        if miles > 0:
            self.__mileage += miles
            print(f"Drove {miles} miles.")
        else:
            print("You can't drive backwards to reduce mileage!")

    def get_mileage(self):
        return f"Current mileage: {self.__mileage}"

my_car = Car("Toyota")
my_car.drive(50)

# This will FAIL or create a new public variable, but won't change the private one:
my_car.__mileage = 0

print(my_car.get_mileage()) # Still 50!
```
